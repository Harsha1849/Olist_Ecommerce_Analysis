```python
customers = pd.read_csv("/content/olist_customers_dataset.csv")
order_items = pd.read_csv("/content/olist_order_items_dataset.csv")
payments = pd.read_csv("/content/olist_order_payments_dataset.csv")
products = pd.read_csv("/content/olist_products_dataset.csv")
reviews = pd.read_csv("/content/olist_order_reviews_dataset.csv")
sellers = pd.read_csv("/content/olist_sellers_dataset.csv")
```

```python
tables = {
    "customers": customers,
    "orders": orders,
    "order_items": order_items,
    "payments": payments,
    "products": products,
    "reviews": reviews,
    "sellers": sellers
}

for name, df in tables.items():
    print(name, df.shape)
```

```python
order_value = (
    order_items.groupby("order_id")
    .agg(
        product_value=("price", "sum"),
        freight_value=("freight_value", "sum"),
        item_count=("order_id", "count")
    )
    .reset_index()
)

order_value.head()
```

```python
orders_analysis = orders.merge(
    order_value,
    on="order_id",
    how="left"
)
```

```python
customer_info = customers[
    ["customer_id", "customer_unique_id", "customer_city", "customer_state"]
]
```

```python
state_orders = (
    orders_analysis
    .groupby("customer_state")
    .agg(
        orders=("order_id", "nunique"),
        customers=("customer_unique_id", "nunique"),
        revenue=("product_value", "sum")
    )
    .sort_values("revenue", ascending=False)
)

state_orders.head(10)
```

```python
order_payments = (
    payments.groupby("order_id")
    .agg(
        payment_value=("payment_value", "sum"),
        payment_installments=("payment_installments", "max"),
        payment_count=("payment_type", "count")
    )
    .reset_index()
)
```

```python
order_payments["order_id"].duplicated().sum()
```

```python
orders_analysis = orders_analysis.merge(
    order_payments,
    on="order_id",
    how="left"
)
```

```python
orders_analysis.head()
```

```python
payment_method_analysis = (
    payments.groupby("payment_type")
    .agg(
        orders=("order_id", "nunique"),
        payment_value=("payment_value", "sum")
    )
    .sort_values("payment_value", ascending=False)
)

payment_method_analysis
```

```python
orders_analysis["payment_value"].mean()
```

```python
unique_customers = orders_analysis["customer_unique_id"].nunique()

print("Unique customers:", unique_customers)
```

```python
customer_orders = (
    orders_analysis
    .groupby("customer_unique_id")
    .agg(
        order_count=("order_id", "nunique"),
        total_spent=("payment_value", "sum")
    )
    .reset_index()
)
```

```python
customer_orders["customer_type"] = customer_orders["order_count"].apply(
    lambda x: "Repeat" if x > 1 else "One-time"
)
```

```python
customer_orders["customer_type"].value_counts()
```

```python
repeat_percentage = (
    (customer_orders["customer_type"] == "Repeat").mean() * 100
)

print(f"Repeat customer rate: {repeat_percentage:.2f}%")
```

```python
top_customers = (
    customer_orders
    .sort_values("total_spent", ascending=False)
    .head(10)
)

top_customers
```

```python
delivery_analysis = orders_analysis.copy()
```

```python
delivery_analysis["delivery_days"] = (
    delivery_analysis["order_delivered_customer_date"]
    - delivery_analysis["order_purchase_timestamp"]
).dt.total_seconds() / (60 * 60 * 24)
```

```python
delivery_analysis["delivery_days"].describe()
```

```python
delivery_analysis["late_delivery"] = (
    delivery_analysis["order_delivered_customer_date"]
    > delivery_analysis["order_estimated_delivery_date"]
)
```

```python
delivery_analysis["late_delivery"].value_counts()
```

```python
late_rate = delivery_analysis["late_delivery"].mean() * 100

print(f"Late delivery rate: {late_rate:.2f}%")
```
